
# Introduction

This project explores three fundamental concepts in reinforcement learning: **Markov Decision Processes (MDPs)**, **Monte Carlo methods**, and **importance sampling**. Through mathematical analysis and numerical computation, we investigate how optimal decisions can be determined under uncertainty, how value functions can be estimated from sampled experiences, and how expectations can be evaluated using alternative probability distributions.

## 1. Markov Decision Processes

We investigate optimal decision-making in a stochastic environment by computing value functions and determining optimal policies through iterative methods.

The optimal state-value function satisfies the **Bellman optimality equation**:

$$
V^*(s) = \max_{a \in \mathcal{A}(s)}
\sum_{s',r} p(s',r \mid s,a)
\left[r + \gamma V^*(s')\right].
$$

## 2. Monte Carlo Methods

We explore the estimation of value functions using sampled returns and examine how these estimates can support policy improvement and exploration.

The Monte Carlo estimate of a state-value function is given by

$$
V_\pi(s) \approx \frac{1}{N(s)}
\sum_{i=1}^{N(s)} G_i(s),
$$

where $G_i(s)$ denotes the return observed following a visit to state $s$, and $N(s)$ is the number of sampled returns used in the estimate.

## 3. Importance Sampling

We investigate importance sampling as a technique for estimating expectations under a target probability distribution using samples drawn from a different distribution, particularly when direct sampling is difficult or computationally expensive.

The importance sampling identity is

$$
\mathbb{E}_{p}[f(X)]
=
\mathbb{E}_{q}
\left[
\frac{p(X)}{q(X)}f(X)
\right],
$$

provided that $q(x)>0$ wherever $p(x)f(x)$ contributes to the expectation.

# Problem 3 (Importance sampling)

Without importance sampling

```python
# import pi
PI = math.pi
random.seed(42)

#
# Define the PDF function for the t-distribution
def f_t(t):
  return 8 / (3 * math.sqrt(5) * PI * ((1 + (t ** 2 / 5)) ** 3))

# Define the function to generate samples from the t-distribution 
def get_t_samples(n_samples):
  return np.random.standard_t(5, size=n_samples) # with degree of freedom 5

# Define the function to estimate sigma squared
def sigma_est(t):
  return t ** 2

# Generate samples and estimate sigma squared
t_samples = get_t_samples(10000)

# Calculate the estimated sigma squared value using the generated samples
est = sigma_est(t_samples)
```

```python
E_T_squared = np.mean(est)

# print out E[T^2]
print("E_T_squared: ", E_T_squared)
```

```python
# Define the PDF function for the standard normal distribution
def f_normal(x):
  return 1 / math.sqrt(2 * PI) * np.exp(-0.5 * x ** 2)

#Generate samples from the standard normal distribution
def get_normal_samples(n_samples):
  U = np.random.randn(10000)
  return U

# Calculate weights for importance sampling
def calculate_weights(x_samples):
  return f_t(x_samples) / f_normal(x_samples)

# Importance sampling to estimate E[T^2]
def importance_sampling(n_samples):
  # Generate a random number from a standard normal distribution
  normal_samples = get_normal_samples(n_samples)
  weights = calculate_weights(normal_samples)
  t_squared_samples = normal_samples ** 2
  expectation = np.mean(weights * t_squared_samples)
  return expectation

# # Estimate E[T^2] using importance sampling
E_T_squared_imporatnce_sampling = importance_sampling(100000)
print("Estimated E[T^2] using importance sampling:", E_T_squared_imporatnce_sampling)
```

```python
difference = E_T_squared - E_T_squared_imporatnce_sampling
print("The difference between = ",difference)
```

The two values are not that close and it's clear that they are not identical. 

This is expected due to differences in the simulation methods and underlying 

distributions used. Variance in results  is common in Monte Carlo simulations, 

especially with heavy-tailed distributions like the t-distribution.
